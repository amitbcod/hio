# Admin Blade Route Conversion

This notebook will perform a bulk search-and-replace across admin blade templates to convert `operator.*` route references to `admin.*` and update include/layout names.

```python
import os
import re

base_dir = os.path.join(r"c:\wamp64\www\holidaysio", "resources", "views", "admin")

replacements = {
    "route('operator.accommodation.": "route('admin.accommodation.",
    'route("operator.accommodation.': 'route("admin.accommodation.',
    "route('operator.activity.": "route('admin.activity.",
    'route("operator.activity.': 'route("admin.activity.',
    "@include('operator.accommodation.": "@include('admin.accommodation.",
    '@include("operator.accommodation.': '@include("admin.accommodation.',
    "@include('operator.activity.": "@include('admin.activity.",
    '@include("operator.activity.': '@include("admin.activity.',
    "@extends('layouts.app')": "@extends('layouts.admin')",
    '@extends("layouts.app")': '@extends("layouts.admin")',
}

files_updated = []

for dirpath, dirnames, filenames in os.walk(base_dir):
    for filename in filenames:
        if not filename.endswith('.blade.php'):
            continue
        path = os.path.join(dirpath, filename)
        with open(path, 'r', encoding='utf-8') as f:
            text = f.read()

        new_text = text
        for old, new in replacements.items():
            new_text = new_text.replace(old, new)

        if new_text != text:
            with open(path, 'w', encoding='utf-8') as f:
                f.write(new_text)
            files_updated.append(path)

files_updated
```
